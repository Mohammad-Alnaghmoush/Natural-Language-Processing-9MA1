```python
#Mohammad_Abdulrouf_Alnaghmoush
#2230004028
#9M01
import re
from collections import Counter
import pandas as pd

#Load the CSV file
df = pd.read_csv("Apple_tweets.csv")

#Initialize a list to store all extracted hashtags
all_hashtags = []

#Iterate through the 'text' column and extract hashtags
for tweet in df["text"].dropna():
    # Convert text to lowercase for consistency and match regex pattern
    # r'#\w+' matches '#' followed by letters, digits, or underscores
    tags = re.findall(r"#\w+", str(tweet).lower())
    all_hashtags.extend(tags)

#Calculate and print the total number of hashtags found
total_hashtags = len(all_hashtags)
print(f"Total hashtags found: {total_hashtags}")

#Count occurrences and find the top 10 most frequent hashtags
hashtag_counter = Counter(all_hashtags)
top_10_hashtags = hashtag_counter.most_common(10)

#Display the top 10 hashtags in a clean format
print("\nTop 10 Hashtags:")
for rank, (tag, count) in enumerate(top_10_hashtags, start=1):
    print(f"{rank}. {tag} : {count}")
```
