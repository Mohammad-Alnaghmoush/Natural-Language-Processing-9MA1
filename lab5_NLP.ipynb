```python
# Task 1: Cosine Similarity

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Sentences
documents = [
    'This is the first document.',
    'This document is the second document.',
    'And this is the third one.',
    'Is this the first document?'
]

# Convert sentences into TF-IDF vectors
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

# Calculate cosine similarity
similarity_matrix = cosine_similarity(tfidf_matrix)

# Display results
result = pd.DataFrame(
    similarity_matrix,
    index=['Sentence 1', 'Sentence 2', 'Sentence 3', 'Sentence 4'],
    columns=['Sentence 1', 'Sentence 2', 'Sentence 3', 'Sentence 4']
)

result
```

```python
# Task 2: TF-IDF

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

# Sentences
documents = [
    'data science is one of the most important fields of science',
    'this is one of the best data science courses',
    'data scientists analyze data'
]

# Create TF-IDF model
tfidf = TfidfVectorizer()

# Transform sentences into TF-IDF values
tfidf_matrix = tfidf.fit_transform(documents)

# Convert results to DataFrame
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf.get_feature_names_out(),
    index=['Sentence 1', 'Sentence 2', 'Sentence 3']
)

# Display TF-IDF scores
print("TF-IDF Scores:")
display(tfidf_df)

# Display the most important word in each sentence
print("\nMost Important Word in Each Sentence:")

for i, row in tfidf_df.iterrows():
    important_word = row.idxmax()
    score = row.max()
    print(f"{i}: {important_word} -> {score:.4f}")
```

```python
#Task3
import pandas as pd
import re
import nltk
import gensim

from nltk.tokenize import word_tokenize

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

# Load dataset
df = pd.read_csv('simpsons_script_lines.csv')

# Remove missing values
df = df.dropna(subset=['spoken_words']).copy()

# Required preprocessing function
def clean_text(text):
    text = text.lower()
    text = re.sub(r"[0-9]", '', text)
    text = re.sub(r"[)(,”“.’$-]", '', text)
    return text

# Apply preprocessing
df['cleaned_text'] = df['spoken_words'].astype(str).apply(clean_text)

# Tokenization
tokens = []

for text in df['cleaned_text']:
    token = word_tokenize(text)
    tokens.append(token)

# Create Skip-Gram Word2Vec model
Skip_gram_model = gensim.models.Word2Vec(
    tokens,
    min_count=1,
    vector_size=100,
    window=5,
    sg=1
)

print("Word2Vec Skip-Gram model trained successfully.")
print("Vocabulary size:", len(Skip_gram_model.wv))
```

```python
# Task 4: Find Similar Words

print("Words similar to 'homer':")
print(Skip_gram_model.wv.most_similar('homer'))

print("\nWords similar to 'marge':")
print(Skip_gram_model.wv.most_similar('marge'))

print("\nWords similar to 'bart':")
print(Skip_gram_model.wv.most_similar('bart'))
```

```python
# Task 5: Find the Odd Word

# 1
group1 = ['jimbo', 'milhouse', 'kearney']
print("Odd word in group 1:")
print(Skip_gram_model.wv.doesnt_match(group1))

# 2
group2 = ['nelson', 'bart', 'milhouse']
print("\nOdd word in group 2:")
print(Skip_gram_model.wv.doesnt_match(group2))

# 3
group3 = ['homer', 'patty', 'selma']
print("\nOdd word in group 3:")
print(Skip_gram_model.wv.doesnt_match(group3))
```
