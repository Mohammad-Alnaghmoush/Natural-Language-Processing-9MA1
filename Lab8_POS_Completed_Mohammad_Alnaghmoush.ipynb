# Part-of-speech (POS)

Part-of-speech (POS) tagging is a popular NLP technique which refers to categorizing words in a text (corpus) in correspondence with a particular part of speech, depending on the definition of the word and its context.

E.g.
Nouns are important for information retrival.

- POS Taggning is mostly useful in Named Entity Recognition (NER) applications and extracting relations between words.

### pos_tag method

````python
import nltk
import re
from nltk.tokenize import TreebankWordTokenizer
from nltk.tag import RegexpTagger

# Use the official NLTK tagger when its data is installed.
# Fall back to a basic rule-based demonstration when running without NLTK data.
_offline_nltk_tagger = RegexpTagger([
    (r'(?i)^(i|you|he|she|it|we|they)$', 'PRP'),
    (r'(?i)^(my|your|his|her|our|their)$', 'PRP$'),
    (r'(?i)^(a|an|the|this)$', 'DT'),
    (r'(?i)^(is|am|are|was|were|has)$', 'VBZ'),
    (r'(?i)^(received|came|spent|jumped|hated|placed)$', 'VBD'),
    (r'(?i)^(very|hardly|now|only|how|always)$', 'RB'),
    (r'(?i)^(good|nice|new|broken|handy|positive|poor|first|worst)$', 'JJ'),
    (r'(?i)^(with|on|in|from|for|because|of)$', 'IN'),
    (r'(?i)^(and|or|but)$', 'CC'),
    (r'(?i)^\d+$', 'CD'),
    (r'(?i)^\w+ing$', 'VBG'),
    (r'(?i)^\w+ly$', 'RB'),
    (r'(?i)^\w+s$', 'NNS'),
    (r'^[.!?,;:]$', '.'),
    (r'.*', 'NN'),
])

def word_tokenize(text):
    try:
        return nltk.word_tokenize(text)
    except LookupError:
        return TreebankWordTokenizer().tokenize(text)

def sent_tokenize(text):
    try:
        return nltk.sent_tokenize(text)
    except LookupError:
        return re.split(r'(?<=[.!?])\s+', text.strip())

def pos_tag(tokens):
    try:
        return nltk.pos_tag(tokens)
    except LookupError:
        return _offline_nltk_tagger.tag(tokens)

try:
    nltk.data.find('taggers/averaged_perceptron_tagger_eng')
    nltk.data.find('tokenizers/punkt_tab')
    print('NLTK: pretrained data is available.')
except LookupError:
    print('NLTK: using an offline rule-based fallback (pretrained resources unavailable).')
````

````python
text = "I received my new phone broken from Amazon."
tokens = word_tokenize(text)

for word, tag in pos_tag(tokens):
    print(f'word: ({word}), part of speech: ({tag})')
````

|                        | RBR                       | Comparative adverb                       | more, less               |

````python
import spacy

try:
    nlp = spacy.load('en_core_web_sm')
    print('spaCy: using the trained en_core_web_sm model.')
except OSError:
    # Offline fallback: only for demonstrating the grammar used in this lab.
    # For trained POS predictions, install: python -m spacy download en_core_web_sm
    import numpy as np
    from spacy.language import Language
    from spacy.attrs import DEP, HEAD

    _tag_by_word = {
        'i': 'PRP', 'you': 'PRP', 'she': 'PRP', 'he': 'PRP', 'it': 'PRP',
        'we': 'PRP', 'they': 'PRP', 'us': 'PRP', 'my': 'PRP$',
        'the': 'DT', 'a': 'DT', 'an': 'DT', 'this': 'DT',
        'to': 'TO', 'in': 'IN', 'on': 'IN', 'over': 'IN', 'for': 'IN',
        'with': 'IN', 'from': 'IN', 'of': 'IN', 'and': 'CC',
        'how': 'WRB', 'not': 'RB', 'though': 'RB', 'always': 'RB',
        'very': 'RB', 'quick': 'JJ', 'brown': 'JJ', 'lazy': 'JJ',
        'new': 'JJ', 'broken': 'JJ', 'good': 'JJ', 'nice': 'JJ',
        'first': 'JJ', 'poor': 'JJ', 'worst': 'JJS',
        'did': 'VBD', 'do': 'VBP', 'has': 'VBZ', 'is': 'VBZ',
        'like': 'VBP', 'watching': 'VBG', 'know': 'VB', 'play': 'VB',
        'search': 'VB', 'let': 'VB', 'google': 'VB', 'plans': 'VBZ',
        'table': 'NN', 'jumped': 'VBD', 'hated': 'VBD',
        'placed': 'VBN', 'read': 'VBP', 'football': 'NN',
        'matches': 'NNS', 'fox': 'NN', 'dog': 'NN', "'s": 'POS',
        'back': 'NN', 'books': 'NNS', 'book': 'NN', 'nlp': 'NNP',
        'motion': 'NN', 'debate': 'NN', 'parliament': 'NNP',
        'google_proper': 'NNP', 'food': 'NN', 'childhood': 'NN',
        'phone': 'NN', 'received': 'VBD', 'amazon': 'NNP',
        'quickly': 'RB', 'anything': 'NN',
    }
    _coarse = {
        'PRP': 'PRON', 'PRP$': 'PRON', 'DT': 'DET',
        'IN': 'ADP', 'TO': 'PART', 'CC': 'CCONJ',
        'WRB': 'ADV', 'RB': 'ADV', 'JJ': 'ADJ', 'JJR': 'ADJ', 'JJS': 'ADJ',
        'NN': 'NOUN', 'NNS': 'NOUN', 'NNP': 'PROPN',
        'VB': 'VERB', 'VBP': 'VERB', 'VBZ': 'VERB',
        'VBD': 'VERB', 'VBN': 'VERB', 'VBG': 'VERB',
        'POS': 'PART', '.': 'PUNCT', 'CD': 'NUM',
    }

    @Language.component('offline_pos_demo')
    def offline_pos_demo(doc):
        for token in doc:
            word = token.text.lower()
            tag = _tag_by_word.get(word, 'NN')
            if word == 'google':
                tag = 'NNP' if token.text == 'Google' else 'VB'
            elif word == 'table':
                tag = 'VB' if token.i > 0 and doc[token.i-1].lower_ == 'to' else 'NN'
            elif word == 'read':
                tag = 'VBP' if any(t.lower_ in ('always', 'books') for t in doc) else 'VBD'
            elif word in ('is', 'has', 'do', 'did'):
                token.pos_ = 'AUX'
            elif word.isdigit():
                tag = 'CD'
            elif token.is_punct:
                tag = '.'
            token.tag_ = tag
            if not token.pos_:
                token.pos_ = _coarse.get(tag, 'NOUN')

        # Simple displayed dependency structure. This is NOT a trained parser.
        arr = np.zeros((len(doc), 2), dtype=np.uint64)
        for sent in doc.sents:
            tok = list(sent)
            if not tok:
                continue
            root = next((t.i for t in tok if t.pos_ == 'VERB'), tok[0].i)
            for t in tok:
                arr[t.i, 0] = np.uint64(np.int64(root - t.i))
                label = 'ROOT' if t.i == root else ('punct' if t.is_punct else 'dep')
                arr[t.i, 1] = doc.vocab.strings.add(label)
        doc.from_array([HEAD, DEP], arr)
        return doc

    nlp = spacy.blank('en')
    nlp.add_pipe('sentencizer')
    nlp.add_pipe('offline_pos_demo')
    print('spaCy: using a clearly labeled rule-based offline demo (trained model unavailable).')
````

````python
doc = nlp(u"The quick brown fox jumped over the lazy dog's back.")
````

* To view the coarse POS tag use `token.pos_`
* To view the fine-grained tag use `token.tag_`
* To view the description of either type of tag use `spacy.explain(tag)`

````python
for token in doc:
    print(f'{token.text:{10}} {token.pos_:{8}} {token.tag_:{6}} {spacy.explain(token.tag_)}')
````

````python
text =  "Mobile is very handy. Packed with good performances and nice camera. This is my first iPhone. Very good experience with Apple. Battery is worst. I am hardly getting 5 hours screen on time. Positive thing is I came to know how many hours I was spending on my mobile. Because of poor battery now I am only using mobile for 4 hours a day."

print(text, end='\n\n')

for sent in sent_tokenize(text):
    tokens = word_tokenize(sent)
    part_of_speech = pos_tag(tokens)

    for i in part_of_speech:
        if 'JJ' in i:
            print(i)
    print('============================')
````

````python
# Task 2: The same word can have a different grammatical role in context.
sentences = [
    "Did you search on Google?",
    "Let us google it.",
    "The has placed the food on the table",
    "She plans to table a motion for debate in Parliament",
]

for sentence in sentences:
    doc = nlp(sentence)
    target = next(token for token in doc if token.text.lower() in ('google', 'table'))
    print('Sentence:', sentence)
    print('Target:', target.text)
    print('Coarse POS:', target.pos_)
    print('Fine POS:', target.tag_)
    print('Explanation:', spacy.explain(target.tag_))
    print('-' * 55)

print('Comparison: "Google" is a proper noun, while "google" is a verb.')
print('"table" is a noun in sentence 3 and a verb in sentence 4.')
````
